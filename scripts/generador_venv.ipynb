# Script para generar el entorno virtual necesario para la ejecución del resto de scripts
## Es necesario debido a la necesidad de utilizar versiones exactas de cada librería

El primer paso es instalar la versión 3.11 de Python. Para ello se puede utilizar este link: https://www.python.org/downloads/

Después, ejecutamos la siguiente celda, que si nos hemos instalado la versión 3.11 de Python debería de funcionar sin problema.
Esto nos  crea el entorno virtual en la carpeta principal del proyecto.


```python
!py -3.11 -m venv ../venv_311
```

Ahora procedemos a instalar las diferentes versiones que necesitamos.

```python
!"../venv_311/Scripts/pip" install torch==2.3.1 torchaudio==2.3.1 torchvision==0.18.1 --index-url https://download.pytorch.org/whl/cpu
!"../venv_311/Scripts/pip" install torchcodec
!"../venv_311/Scripts/pip" install "numpy<2" --force-reinstall
```

Por último registramos el Kernel

```python
!"../venv_pyo_311/Scripts/python" -m ipykernel install --user --name venv_311 --display-name "Python 3.11 (venv) script"
```
